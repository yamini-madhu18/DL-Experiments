```python
print("Yamini M")
print("24BAD131")

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

(x_train, _), (x_test, _) = tf.keras.datasets.mnist.load_data()

print("Training images shape:", x_train.shape)
print("Testing images shape:", x_test.shape)
print("Image dimensions:", x_train.shape[1:])

plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap='gray')
    plt.axis('off')
plt.show()

x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

x_train = x_train.reshape((len(x_train), 784))
x_test = x_test.reshape((len(x_test), 784))

print("After preprocessing:")
print("Training data shape:", x_train.shape)
print("Testing data shape:", x_test.shape)
print("Pixel value range:", x_train.min(), "to", x_train.max())
```

```python
print("Yamini M")
print("24BAD131")

from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Model

input_img = Input(shape=(784,))

encoded = Dense(128, activation='relu')(input_img)
latent = Dense(32, activation='relu')(encoded)

decoded = Dense(128, activation='relu')(latent)
output_img = Dense(784, activation='sigmoid')(decoded)

autoencoder = Model(input_img, output_img)

encoder = Model(input_img, latent)

decoder_input = Input(shape=(32,))
decoder_layer = autoencoder.layers[-2](decoder_input)
decoder_output = autoencoder.layers[-1](decoder_layer)
decoder = Model(decoder_input, decoder_output)

print("Encoder Summary:")
encoder.summary()

print("Decoder Summary:")
decoder.summary()

print("Autoencoder Summary:")
autoencoder.summary()
```

```python
print("Yamini M")
print("24BAD131")

autoencoder.compile(optimizer='adam', loss='binary_crossentropy')

history = autoencoder.fit(
    x_train, x_train,
    epochs=10,
    batch_size=256,
    shuffle=True,
    validation_data=(x_test, x_test)
)

plt.figure(figsize=(8, 5))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Autoencoder Training and Validation Loss')
plt.legend()
plt.show()

print("Final Training Loss:", history.history['loss'][-1])
print("Final Validation Loss:", history.history['val_loss'][-1])
```

```python
print("Yamini M")
print("24BAD131")

latent_features = encoder.predict(x_test)
reconstructed_images = autoencoder.predict(x_test)

print("Original Input Dimension:", x_test.shape[1])
print("Latent Dimension:", latent_features.shape[1])
print("Latent Feature Shape:", latent_features.shape)

plt.figure(figsize=(12, 5))

for i in range(10):
    plt.subplot(2, 10, i + 1)
    plt.imshow(x_test[i].reshape(28, 28), cmap='gray')
    plt.axis('off')
    if i == 0:
        plt.ylabel("Original")

    plt.subplot(2, 10, i + 11)
    plt.imshow(reconstructed_images[i].reshape(28, 28), cmap='gray')
    plt.axis('off')
    if i == 0:
        plt.ylabel("Reconstructed")

plt.show()

print("Original Dimension: 784")
print("Reduced Dimension: 32")
print("Dimensionality Reduction Completed Successfully")

autoencoder.save("autoencoder_model.keras")
encoder.save("encoder_model.keras")
```
